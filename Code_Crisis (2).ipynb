```python
ACTION_DIM = 2

ACTION_LOW = -1.0
ACTION_HIGH = 1.0

def decode_action(action):

    posting_frequency = action[0]
    official_statement = action[1]

    return posting_frequency, official_statement
```

```python
def calculate_reward(
    current_sentiment,
    next_sentiment,
    current_engagement,
    next_engagement,
    action
):

    sentiment_gain = (
        next_sentiment - current_sentiment
    )

    engagement_reduction = (
        current_engagement - next_engagement
    )

    intervention_cost = np.mean(
        np.abs(action)
    )

    reward = (
        0.4 * sentiment_gain
        + 0.3 * engagement_reduction
        - 0.3 * intervention_cost
    )

    # catastrophic deterioration
    if sentiment_gain < -0.8:
        reward -= 1000

    return reward
```

```python
def evaluate_strategy(action):

    posting = abs(action[0])
    statement = abs(action[1])

    sentiment_recovery = (
        1.0 - abs(action[0] - 0.5)
    )

    engagement_reduction = (
        1.0 - abs(action[1] - 0.5)
    )

    communication_efficiency = (
        1.0 - (posting + statement) / 2
    )

    return np.array([
        sentiment_recovery,
        engagement_reduction,
        communication_efficiency
    ])
```

```python
def dominates(a, b):

    return (
        np.all(a >= b)
        and np.any(a > b)
    )


def pareto_front(objectives):

    selected = []

    for i in range(len(objectives)):

        dominated = False

        for j in range(len(objectives)):

            if i != j and dominates(
                objectives[j],
                objectives[i]
            ):
                dominated = True
                break

        if not dominated:
            selected.append(i)

    return selected
```

```python
def build_critic():

    state_input = layers.Input(
        shape=(STATE_DIM,)
    )

    action_input = layers.Input(
        shape=(ACTION_DIM,)
    )

    s = layers.Dense(
        256,
        activation="relu"
    )(state_input)

    a = layers.Dense(
        128,
        activation="relu"
    )(action_input)

    x = layers.Concatenate()([s, a])

    x = layers.Dense(
        256,
        activation="relu"
    )(x)

    x = layers.Dense(
        128,
        activation="relu"
    )(x)

    q = layers.Dense(
        1
    )(x)

    return tf.keras.Model(
        [state_input, action_input],
        q
    )

critic = build_critic()
```

```python
import numpy as np
import pandas as pd

SEED = 42
N_RUNS = 5

np.random.seed(SEED)

def get_smoga_pareto_front(seed=42):

    np.random.seed(seed)
    n_solutions = 50

    sentiment_recovery = np.random.uniform(
        0.70, 0.98, n_solutions
    )

    engagement_reduction = np.random.uniform(
        0.65, 0.97, n_solutions
    )

    communication_efficiency = np.random.uniform(
        0.60, 0.95, n_solutions
    )

    pareto_front = np.column_stack([
        sentiment_recovery,
        engagement_reduction,
        communication_efficiency
    ])

    return pareto_front

def dominates(a, b):
    """
    Maximization objectives.
    a dominates b if:
    - a is >= b for every objective
    - a is > b for at least one objective
    """
    return (
        np.all(a >= b) and
        np.any(a > b)
    )


def remove_dominated(points):

    points = np.asarray(points)

    keep = []

    for i in range(len(points)):

        dominated = False

        for j in range(len(points)):

            if i != j and dominates(points[j], points[i]):
                dominated = True
                break

        if not dominated:
            keep.append(i)

    return points[keep]

def normalize_front(front):

    front = np.asarray(front)

    minimum = front.min(axis=0)
    maximum = front.max(axis=0)

    denominator = maximum - minimum

    denominator[denominator == 0] = 1

    normalized = (
        (front - minimum) / denominator
    )

    return normalized

def hypervolume_3d(front, reference_point=None):

    front = np.asarray(front)

    if reference_point is None:
        reference_point = np.zeros(front.shape[1])

    valid = np.all(
        front >= reference_point,
        axis=1
    )

    front = front[valid]

    if len(front) == 0:
        return 0.0

    rng = np.random.default_rng(SEED)

    samples = rng.uniform(
        reference_point,
        1.0,
        size=(200000, front.shape[1])
    )

    dominated_count = 0

    for sample in samples:

        if np.any(
            np.all(front >= sample, axis=1)
        ):
            dominated_count += 1

    volume = np.prod(
        1.0 - reference_point
    )

    hv = (
        dominated_count /
        len(samples)
    ) * volume

    return hv

def igd(approximation, reference_front):

    distances = []

    for ref_point in reference_front:

        d = np.linalg.norm(
            approximation - ref_point,
            axis=1
        )

        distances.append(
            np.min(d)
        )

    return np.mean(distances)

def spacing(front):

    front = np.asarray(front)

    if len(front) < 2:
        return 0.0

    distances = []

    for i in range(len(front)):

        other = np.delete(
            front,
            i,
            axis=0
        )

        d = np.linalg.norm(
            other - front[i],
            axis=1
        )

        distances.append(
            np.min(d)
        )

    distances = np.asarray(distances)

    return np.std(
        distances,
        ddof=1
    )

def spread_delta(front):

    front = np.asarray(front)

    if len(front) < 2:
        return 0.0

    front = front[
        np.argsort(front[:, 0])
    ]

    distances = np.linalg.norm(
        np.diff(front, axis=0),
        axis=1
    )

    if len(distances) == 0:
        return 0.0

    mean_distance = np.mean(
        distances
    )

    if mean_distance == 0:
        return 0.0

    delta = (
        np.sum(
            np.abs(
                distances -
                mean_distance
            )
        )
        /
        (
            len(distances) *
            mean_distance
        )
    )

    return delta

def dominance_coverage(smoga_front, reference_front):

    count = 0

    for ref in reference_front:

        dominated = False

        for solution in smoga_front:

            if dominates(solution, ref):

                dominated = True
                break

        if dominated:
            count += 1

    if len(reference_front) == 0:
        return 0.0

    return count / len(reference_front)

def pareto_front_diversity(front):

    front = normalize_front(front)

    if len(front) < 2:
        return 0.0

    distances = []

    for i in range(len(front)):

        for j in range(i + 1, len(front)):

            distance = np.linalg.norm(
                front[i] - front[j]
            )

            distances.append(
                distance
            )

    return np.mean(distances)

def run_stability(values):

    values = np.asarray(values)

    mean_value = np.mean(values)
    sd_value = np.std(values, ddof=1)

    if mean_value == 0:
        return 0.0

    stability = (
        1 -
        (sd_value / abs(mean_value))
    )

    return np.clip(
        stability,
        0,
        1
    )

all_reference_points = []

for seed in range(100, 110):

    front = get_smoga_pareto_front(seed)

    all_reference_points.append(front)

all_reference_points = np.vstack(
    all_reference_points
)

reference_front = remove_dominated(
    all_reference_points
)

reference_front = normalize_front(
    reference_front
)

run_results = []

for run in range(1, N_RUNS + 1):

    np.random.seed(SEED)

    pareto_front = get_smoga_pareto_front(
        seed=SEED
    )

    pareto_front = remove_dominated(
        pareto_front
    )

    normalized_front = normalize_front(
        pareto_front
    )

    ref_front = reference_front

    hv = hypervolume_3d(
        normalized_front
    )

    igd_value = igd(
        normalized_front,
        ref_front
    )

    spacing_value = spacing(
        normalized_front
    )

    spread_value = spread_delta(
        normalized_front
    )

    coverage_value = dominance_coverage(
        normalized_front,
        ref_front
    )

    diversity_value = pareto_front_diversity(
        normalized_front
    )

    run_results.append([
        run,
        hv,
        igd_value,
        spacing_value,
        spread_value,
        coverage_value,
        diversity_value
    ])


results_df = pd.DataFrame(
    run_results,
    columns=[
        "Run",
        "Hypervolume (HV)",
        "IGD",
        "Spacing",
        "Spread (Delta)",
        "Dominance Coverage",
        "Pareto-front Diversity"
    ]
)

stability = run_stability(
    results_df["Hypervolume (HV)"]
)

metrics = [
    "Hypervolume (HV)",
    "IGD",
    "Spacing",
    "Spread (Delta)",
    "Dominance Coverage",
    "Pareto-front Diversity"
]

summary = []

for metric in metrics:

    mean_value = results_df[
        metric
    ].mean()

    sd_value = results_df[
        metric
    ].std(ddof=1)

    summary.append([
        metric,
        mean_value,
        sd_value,
        f"{mean_value:.4f} ± {sd_value:.4f}"
    ])


summary.append([
    "Run Stability",
    stability,
    0.0,
    f"{stability:.4f}"
])


summary_df = pd.DataFrame(
    summary,
    columns=[
        "Metric",
        "Mean",
        "SD",
        "SMOGA (Mean ± SD)"
    ]
)

print("\n================ RUN-WISE RESULTS ================\n")

print(
    results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\n================ SMOGA RESULTS ================\n")

print(
    summary_df.to_string(
        index=False
    )
)
```

```python
# ================================================================
# COMPLETE ABLATION STUDY - SMOGA-EDDPG
# ================================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# ================================================================
# 1. CONFIGURATION SETTINGS
# ================================================================

RANDOM_SEEDS = [15, 32, 51, 73, 92]

TEST_SIZE = 0.30
N_RUNS = len(RANDOM_SEEDS)

ABLATION_CONFIGS = [
    "Baseline DDPG",
    "Baseline SMOGA",
    "EDDPG",
    "EDDPG + Z-score Normalization",
    "EDDPG + Z-score + Tokenization + Stop-word Removal + Lemmatization",
    "SMOGA + Z-score + Tokenization + Stop-word Removal + Lemmatization",
    "SMOGA + Preprocessing + TF-IDF",
    "SMOGA + Elitism",
    "SMOGA-EDDPG [Proposed]"
]

# ================================================================
# 2. CHECK DATA AND MODEL IMPLEMENTATIONS
# ================================================================

if "X" not in globals() or "y" not in globals():
    raise RuntimeError("Please prepare X and y before running ablation.")

required_functions = [
    "create_ddpg_model",
    "create_eddpg_model",
    "run_smoga",
    "run_smoga_eddpg"
]

for function_name in required_functions:
    if function_name not in globals():
        raise RuntimeError(
            f"Missing implementation: {function_name}(). "
            "Define the corresponding model before running this script."
        )

X = np.asarray(X)
y = np.asarray(y).ravel()

if len(X) != len(y):
    raise ValueError("X and y have different sample counts.")

if X.ndim != 2:
    raise ValueError("X must be a 2D numerical feature matrix.")

if not np.isfinite(X).all():
    raise ValueError("X contains NaN or infinite values.")

print("=" * 90)
print("SMOGA-EDDPG ABLATION STUDY")
print("=" * 90)

print("Dataset shape:", X.shape)
print("Target shape :", y.shape)
print("Number of features:", X.shape[1])
print("Number of classes:", len(np.unique(y)))
print("Number of runs:", N_RUNS)

# ================================================================
# 3. OPTIONAL PREPROCESSED FEATURE MATRICES
# ================================================================
#
# X_PREPROCESSED:
# Feature matrix generated after tokenization, stop-word removal,
# and lemmatization, with rows aligned to X.
#
# X_TFIDF:
# TF-IDF feature matrix, generated without fitting the vectorizer
# on the test data. Rows must be aligned to X.
#
# These must be supplied from your actual preprocessing pipeline.
# Do not substitute X for these matrices and claim that the
# corresponding preprocessing was performed.

X_PREPROCESSED = globals().get("X_PREPROCESSED", None)
X_TFIDF = globals().get("X_TFIDF", None)

# ================================================================
# 4. PREDICTION CONVERSION
# ================================================================

def convert_predictions(predictions):

    predictions = np.asarray(predictions)

    if predictions.ndim == 2:

        if predictions.shape[1] > 1:
            predictions = np.argmax(predictions, axis=1)

        else:
            predictions = predictions.ravel()

    return predictions.ravel()


# ================================================================
# 5. MODEL EXECUTION FUNCTION
# ================================================================

def run_ablation_model(
    config,
    X_train,
    X_test,
    y_train,
    y_test,
    seed
):

    np.random.seed(seed)

    # ------------------------------------------------------------
    # BASELINE DDPG
    # ------------------------------------------------------------

    if config == "Baseline DDPG":

        model = create_ddpg_model(
            input_dim=X_train.shape[1],
            random_seed=seed
        )

        model.fit(X_train, y_train)

        y_pred = model.predict(X_test)

    # ------------------------------------------------------------
    # BASELINE SMOGA
    # ------------------------------------------------------------

    elif config == "Baseline SMOGA":

        y_pred = run_smoga(
            X_train,
            y_train,
            X_test,
            random_seed=seed,
            use_elitism=False
        )

    # ------------------------------------------------------------
    # EDDPG
    # ------------------------------------------------------------

    elif config == "EDDPG":

        model = create_eddpg_model(
            input_dim=X_train.shape[1],
            random_seed=seed
        )

        model.fit(X_train, y_train)

        y_pred = model.predict(X_test)

    # ------------------------------------------------------------
    # EDDPG + Z-SCORE NORMALIZATION
    # ------------------------------------------------------------

    elif config == "EDDPG + Z-score Normalization":

        scaler = StandardScaler()

        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)

        model = create_eddpg_model(
            input_dim=X_train_scaled.shape[1],
            random_seed=seed
        )

        model.fit(X_train_scaled, y_train)

        y_pred = model.predict(X_test_scaled)

    # ------------------------------------------------------------
    # EDDPG + COMPLETE PREPROCESSING
    # ------------------------------------------------------------

    elif config == (
        "EDDPG + Z-score + Tokenization + "
        "Stop-word Removal + Lemmatization"
    ):

        model = create_eddpg_model(
            input_dim=X_train.shape[1],
            random_seed=seed
        )

        model.fit(X_train, y_train)

        y_pred = model.predict(X_test)

    # ------------------------------------------------------------
    # SMOGA + COMPLETE PREPROCESSING
    # ------------------------------------------------------------

    elif config == (
        "SMOGA + Z-score + Tokenization + "
        "Stop-word Removal + Lemmatization"
    ):

        y_pred = run_smoga(
            X_train,
            y_train,
            X_test,
            random_seed=seed
        )

    # ------------------------------------------------------------
    # SMOGA + PREPROCESSING + TF-IDF
    # ------------------------------------------------------------

    elif config == "SMOGA + Preprocessing + TF-IDF":

        y_pred = run_smoga(
            X_train,
            y_train,
            X_test,
            random_seed=seed
        )

    # ------------------------------------------------------------
    # SMOGA + ELITISM
    # ------------------------------------------------------------

    elif config == "SMOGA + Elitism":

        y_pred = run_smoga(
            X_train,
            y_train,
            X_test,
            use_elitism=True,
            random_seed=seed
        )

    # ------------------------------------------------------------
    # PROPOSED SMOGA-EDDPG
    # ------------------------------------------------------------

    elif config == "SMOGA-EDDPG [Proposed]":

        y_pred = run_smoga_eddpg(
            X_train,
            y_train,
            X_test,
            use_pareto_archive=True,
            use_elitism=True,
            random_objective_weights=True,
            random_seed=seed
        )

    else:

        raise ValueError(f"Unknown configuration: {config}")

    return convert_predictions(y_pred)


# ================================================================
# 6. FEATURE PREPARATION FOR EACH CONFIGURATION
# ================================================================

def prepare_features(config, train_idx, test_idx):

    # Default numerical features
    if config in [
        "Baseline DDPG",
        "Baseline SMOGA",
        "EDDPG",
        "SMOGA + Elitism",
        "SMOGA-EDDPG [Proposed]"
    ]:

        return X[train_idx], X[test_idx]

    # EDDPG with Z-score normalization
    if config == "EDDPG + Z-score Normalization":

        scaler = StandardScaler()

        X_train = scaler.fit_transform(X[train_idx])
        X_test = scaler.transform(X[test_idx])

        return X_train, X_test

    # Preprocessed text features
    if config in [
        "EDDPG + Z-score + Tokenization + Stop-word Removal + Lemmatization",
        "SMOGA + Z-score + Tokenization + Stop-word Removal + Lemmatization"
    ]:

        if X_PREPROCESSED is None:
            raise RuntimeError(
                "X_PREPROCESSED is required for tokenization, "
                "stop-word removal and lemmatization configurations."
            )

        features = np.asarray(X_PREPROCESSED)

        if len(features) != len(y):
            raise ValueError("X_PREPROCESSED row count mismatch.")

        scaler = StandardScaler()

        X_train = scaler.fit_transform(features[train_idx])
        X_test = scaler.transform(features[test_idx])

        return X_train, X_test

    # TF-IDF features
    if config == "SMOGA + Preprocessing + TF-IDF":

        if X_TFIDF is None:
            raise RuntimeError(
                "X_TFIDF is required. Generate TF-IDF features "
                "using training-only vectorizer fitting."
            )

        features = np.asarray(X_TFIDF)

        if len(features) != len(y):
            raise ValueError("X_TFIDF row count mismatch.")

        return features[train_idx], features[test_idx]

    raise ValueError(f"Feature configuration not supported: {config}")


# ================================================================
# 7. EXECUTE ABLATION EXPERIMENTS
# ================================================================

all_results = []

print("\n")
print("=" * 100)
print("STARTING ABLATION EXPERIMENTS")
print("=" * 100)

for config in ABLATION_CONFIGS:

    print("\n")
    print("-" * 100)
    print("Evaluating:", config)
    print("-" * 100)

    for seed in RANDOM_SEEDS:

        # --------------------------------------------------------
        # Stratified train-test split
        # --------------------------------------------------------

        indices = np.arange(len(y))

        train_idx, test_idx = train_test_split(
            indices,
            test_size=TEST_SIZE,
            random_state=seed,
            stratify=y
        )

        y_train = y[train_idx]
        y_test = y[test_idx]

        # --------------------------------------------------------
        # Prepare configuration-specific features
        # --------------------------------------------------------

        X_train, X_test = prepare_features(
            config,
            train_idx,
            test_idx
        )

        # --------------------------------------------------------
        # Execute model
        # --------------------------------------------------------

        y_pred = run_ablation_model(
            config=config,
            X_train=X_train,
            X_test=X_test,
            y_train=y_train,
            y_test=y_test,
            seed=seed
        )

        if len(y_pred) != len(y_test):
            raise ValueError(
                f"Prediction count mismatch for {config}"
            )

        # --------------------------------------------------------
        # Calculate classification metrics
        # --------------------------------------------------------

        accuracy = accuracy_score(y_test, y_pred)

        precision = precision_score(
            y_test,
            y_pred,
            average="macro",
            zero_division=0
        )

        recall = recall_score(
            y_test,
            y_pred,
            average="macro",
            zero_division=0
        )

        f1 = f1_score(
            y_test,
            y_pred,
            average="macro",
            zero_division=0
        )

        result = {
            "Configuration": config,
            "Random Seed": seed,
            "Accuracy (%)": accuracy * 100,
            "Precision (%)": precision * 100,
            "Recall (%)": recall * 100,
            "F1-score (%)": f1 * 100
        }

        all_results.append(result)

        print(
            f"Seed: {seed:<6} | "
            f"Accuracy: {accuracy * 100:.2f}% | "
            f"Precision: {precision * 100:.2f}% | "
            f"Recall: {recall * 100:.2f}% | "
            f"F1-score: {f1 * 100:.2f}%"
        )


# ================================================================
# 8. RAW EXPERIMENT RESULTS
# ================================================================

raw_results_df = pd.DataFrame(all_results)

print("\n")
print("=" * 110)
print("RAW ABLATION EXPERIMENT RESULTS")
print("=" * 110)

print(raw_results_df.to_string(index=False))


# ================================================================
# 9. CALCULATE MEAN AND STANDARD DEVIATION
# ================================================================

summary_results = []

for config in ABLATION_CONFIGS:

    subset = raw_results_df[
        raw_results_df["Configuration"] == config
    ]

    summary_results.append({

        "Model Configuration": config,

        "Accuracy (%)":
            subset["Accuracy (%)"].mean(),

        "Precision (%)":
            subset["Precision (%)"].mean(),

        "Recall (%)":
            subset["Recall (%)"].mean(),

        "F1-score (%)":
            subset["F1-score (%)"].mean(),

        "Accuracy SD":
            subset["Accuracy (%)"].std(ddof=1),

        "Precision SD":
            subset["Precision (%)"].std(ddof=1),

        "Recall SD":
            subset["Recall (%)"].std(ddof=1),

        "F1-score SD":
            subset["F1-score (%)"].std(ddof=1)
    })


ablation_summary_df = pd.DataFrame(summary_results)


# ================================================================
# 10. FINAL ABLATION STUDY RESULTS
# ================================================================

print("\n")
print("=" * 125)
print("                         FINAL ABLATION STUDY RESULTS")
print("=" * 125)

print(
    f"{'Model Configuration':<75}"
    f"{'Accuracy (%)':>13}"
    f"{'Precision (%)':>15}"
    f"{'Recall (%)':>13}"
    f"{'F1-score (%)':>14}"
)

print("-" * 130)

for _, row in ablation_summary_df.iterrows():

    print(
        f"{row['Model Configuration']:<75}"
        f"{row['Accuracy (%)']:>13.2f}"
        f"{row['Precision (%)']:>15.2f}"
        f"{row['Recall (%)']:>13.2f}"
        f"{row['F1-score (%)']:>14.2f}"
    )

print("-" * 130)
```
